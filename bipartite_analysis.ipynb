```python
import networkx as nx
from networkx.algorithms import bipartite
import community as community_louvain
import pandas as pd
```

1. Memuat file bipartite_edges.csv

```python
edges_df = pd.read_csv('src/data/bipartite_edges.csv')
```

 Membuat graph kosong

```python
B = nx.Graph()
```

2. Menambahkan edge (sisi) dan node dengan atribut bipartite

```python
for index, row in edges_df.iterrows():
    B.add_node(row['UserId'], bipartite=0)
    B.add_node(row['ProductId'], bipartite=1)
    B.add_edge(row['UserId'], row['ProductId'])

print(f"Jumlah Total Node: {B.number_of_nodes()}")
print(f"Jumlah Total Edge: {B.number_of_edges()}")
```

3. Menentukan dua kelompok node


```python
user_nodes = {n for n, d in B.nodes(data=True) if d.get('bipartite') == 0}
product_nodes = {n for n, d in B.nodes(data=True) if d.get('bipartite') == 1}

print(f"Jumlah User Nodes (bipartite=0): {len(user_nodes)}")
print(f"Jumlah Product Nodes (bipartite=1): {len(product_nodes)}")
```

4. Proyeksi Graf & Community Detection dengan Algoritma Louvain

```python
user_projected = bipartite.projected_graph(B, user_nodes)
partition = community_louvain.best_partition(user_projected)
```

Menampilkan beberapa hasil komunitas pertama

```python
print("\nContoh hasil deteksi komunitas (User -> Nomor Komunitas):")
count = 0
for user, comm_id in partition.items():
    print(f"User: {user} -> Komunitas: {comm_id}")
    count += 1
    if count >= 5:
        break
```
